DORA is the nickname for the four‑message handshake DHCP uses to assign an IP address.

| Step | Message | What happens |
|------|---------|--------------|
| D | Discover | The client broadcasts a DHCPDISCOVER to find any DHCP server. |
| O | Offer | A server replies with a DHCPOFFER, proposing an IP address and lease parameters. |
| R | Request | The client selects an offer and sends a DHCPREQUEST to claim that address. |
| A | Acknowledge | The server confirms with a DHCPACK, finalizing the lease (or sends DHCPNAK if it can’t). |

So DORA = Discover → Offer → Request → Acknowledge, the core DHCP lease acquisition process.
